```python
import numpy as np
import pandas as pd
lotto = pd.read_csv('C:\\lottery\\649\\649.csv', index_col = 'DATE')
```

```python
import warnings
warnings.filterwarnings("ignore")
def merge(df,m):
    sum_n=pd.DataFrame(columns=["number","sum"])
    for i in range(1,50):
        sum_n=sum_n.append({'number': i, 'sum': 0}, ignore_index=True)
    df1=df['NUMBER DRAWN 1'][-m:].value_counts().to_frame().reset_index()
    df2=df['NUMBER DRAWN 2'][-m:].value_counts().to_frame().reset_index() 
    df3=df['NUMBER DRAWN 3'][-m:].value_counts().to_frame().reset_index()
    df4=df['NUMBER DRAWN 4'][-m:].value_counts().to_frame().reset_index()
    df5=df['NUMBER DRAWN 5'][-m:].value_counts().to_frame().reset_index()
    df6=df['NUMBER DRAWN 6'][-m:].value_counts().to_frame().reset_index()
    df7=df['BONUS NUMBER'][-m:].value_counts().to_frame().reset_index()
    for i in range(0,df1.shape[0]):
        sum_n.iloc[df1.iloc[i,0]-1,1]+=df1.iloc[i,1]
    for i in range(0,df2.shape[0]):
        sum_n.iloc[df2.iloc[i,0]-1,1]+=df2.iloc[i,1]
    for i in range(0,df3.shape[0]):
        sum_n.iloc[df3.iloc[i,0]-1,1]+=df3.iloc[i,1]
    for i in range(0,df4.shape[0]):
        sum_n.iloc[df4.iloc[i,0]-1,1]+=df4.iloc[i,1]
    for i in range(0,df5.shape[0]):
        sum_n.iloc[df5.iloc[i,0]-1,1]+=df5.iloc[i,1]
    for i in range(0,df6.shape[0]):
        sum_n.iloc[df6.iloc[i,0]-1,1]+=df6.iloc[i,1]
    for i in range(0,df7.shape[0]):
        sum_n.iloc[df7.iloc[i,0]-1,1]+=df7.iloc[i,1]
    sum_n=sum_n.sort_values(by=["sum"])
    return sum_n
s1=merge(lotto,57)
s2=merge(lotto,189)
s3=s1.head(2).append(s2.tail(2))
s4=s1.iloc[[23]].append(s2.iloc[[24]])
s5=s3.append(s4)
print(s5.iloc[:,0].to_list())
```
